```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_digits
```

```python
dataset=load_digits()
```

```python
dataset
```

```python
data=pd.DataFrame(dataset["data"],columns=dataset["feature_names"])
```

```python
data
```

```python
dataset["data"][0].reshape(8,8)
```

```python
#image data
plt.matshow(dataset["data"][0].reshape(8,8),cmap="Greens")
```

```python
#image data
plt.matshow(dataset["data"][1796].reshape(8,8),cmap="Greens")
```

```python
# Standardization of data
from sklearn.preprocessing  import StandardScaler
```

```python
sc=StandardScaler()
```

```python
scaled_X=sc.fit_transform(data)
scaled_X
```

```python
scaled_X=pd.DataFrame(scaled_X,columns=data.columns)
scaled_X
```

```python
# Dependent column
y=dataset["target"]
y
```

```python
y.shape
```

```python
#Implementation of LDA model
```

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
```

```python
y=dataset["target"]
```

```python
pd.Series(y).unique()
#Total 10 Classes
```

```python
lda=LinearDiscriminantAnalysis(n_components=9)
```

```python
lda
```

```python
# Fitting the model
lda_X=lda.fit_transform(scaled_X,y)
```

```python
lda_X
```

```python
lda_X=pd.DataFrame(lda_X)
```

```python
lda_X   # independent
```

```python
y #dependent
```

```python
#splitting the data for training and Testing
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(lda_X,y,train_size=0.7,random_state=46)
```

```python
x_train
```

```python
y_train.shape
```

```python
x_test
```

```python
y_test.shape
```

# **Decision Tree**

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt_model=DecisionTreeClassifier()
```

```python
#Fiting the model
dt_model.fit(x_train,y_train)
```

```python
# testing
dt_pred=dt_model.predict(x_test)
dt_pred
```

```python
y_test
```

```python
from sklearn.metrics import *
```

```python
dt_acc=accuracy_score(y_test,dt_pred)
dt_acc
```

# **Random Forest**

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf_model=RandomForestClassifier()
```

```python
rf_model.fit(x_train,y_train)
```

```python
rf_pred=rf_model.predict(x_test)
rf_pred
```

```python
y_test
```

```python
rf_acc=accuracy_score(y_test,rf_pred)
rf_acc
```

```python
confusion_matrix(y_test,rf_pred)
```

```python
sns.heatmap(confusion_matrix(y_test,rf_pred),cmap="Blues",annot=True)
```

```python
# visualize the random Forest
from sklearn.tree import plot_tree
plt.figure(figsize=(20,20))
plot_tree(rf_model.estimators_[0])
plt.show()
```
