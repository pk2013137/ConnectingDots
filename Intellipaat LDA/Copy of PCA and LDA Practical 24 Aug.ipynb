### **Machine Learning Process**

```python
X = data.drop(columns = 'Play')
y = data['Play']
```

```python
#Splitting the data into Training and Tes sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, stratify = y, random_state = 42)
```

#### Standardization of the data

- To be done after splitting the data always.
- Standaridization should be done only when you are using Linear Algorithms like Linear Regression, Logistic Regression, SVM etc. We don't have to do standaridization when using Tree Based Algorithms.

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

### **Apply Logistic Regression on the data**

```python
log_reg = LogisticRegression()
log_reg.fit(X_train_scaled, y_train)
```

```python
y_pred = log_reg.predict(X_test_scaled)
```

```python
accuracy_score(y_test, y_pred)
```

```python
roc_auc_score(y_test, y_pred)
```

### **Transform the data with PCA**

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components = 0.9)
```

```python
X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)
```

```python
X_train_pca.shape
```

```python
pd.DataFrame(X_train_pca)
```

### **Applying the Logistic Regression on the data which is transformed using PCA**

```python
log_reg2 = LogisticRegression()

log_reg2.fit(X_train_pca, y_train)
```

```python
y_pred2 = log_reg2.predict(X_test_pca)
```

```python
roc_auc_score(y_test, y_pred2)
```
