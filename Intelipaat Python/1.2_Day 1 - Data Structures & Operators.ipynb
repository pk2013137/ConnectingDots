# Properties of Data Structures
1. Ordered or Not - the items are retained in the order in which they are inserted
1. Mutable or Not - Append or Insert Operation permitted on mutable
1. Heterogeneous - Multiple Data types can be stored together
1. Allows duplicates or Not - Removes the duplicates
