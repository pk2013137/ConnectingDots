```python
# prompt: In a 'XYZ' company, the loan department needs to predict if the customer will apply for a housing loan or not. Given past data of customers.
# Determine the overall entropy. for data set df

import pandas as pd
import numpy as np

def calculate_entropy(df, target_column):
    """
    Calculates the overall entropy of a dataset.

    Args:
        df: The input DataFrame.
        target_column: The name of the target column.

    Returns:
        The overall entropy of the dataset.
    """

    # Count the occurrences of each unique value in the target column
    value_counts = df[target_column].value_counts()

    # Calculate the probability of each value
    probabilities = value_counts / len(df)

    # Calculate the entropy for each value
    entropy_values = -probabilities * np.log2(probabilities)

    # Calculate the overall entropy by summing the individual entropies
    overall_entropy = np.sum(entropy_values)

    return overall_entropy


# Assuming 'Housing Loan' is your target variable
overall_entropy = calculate_entropy(df, 'Housing Loan')
print("Overall Entropy:", overall_entropy)
```

```python
# prompt: A manufacturing company is set up in two different locations. If the number of employees in one location are 695, the average monthly salary for their employees is 13875, the number of employees in the other location are 597 and the average monthly salary for their employees is 17086. Find the combined arithmetic mean of the monthly salary.

# Calculate the total salary for each location
total_salary_location1 = 695 * 13875
total_salary_location2 = 597 * 17086

# Calculate the total number of employees
total_employees = 695 + 597

# Calculate the combined total salary
combined_total_salary = total_salary_location1 + total_salary_location2

# Calculate the combined arithmetic mean of the monthly salary
combined_mean_salary = combined_total_salary / total_employees

print(f"The combined arithmetic mean of the monthly salary is: {combined_mean_salary}")
```

```python
# prompt: Which among the following options best describes the Monotonic behavior of y for an increase in x for the given data?  # Sample data
# x = [1.15, -2.16, -1.73, -1.63, 8.79, 2.76, 3.46]
# y = [-4.64, 2.98, 1.99, 1.76, -22.21, -8.34, -9.95]

import numpy as np

# Sample data
x = [1.15, -2.16, -1.73, -1.63, 8.79, 2.76, 3.46]
y = [-4.64, 2.98, 1.99, 1.76, -22.21, -8.34, -9.95]

# Check if y is monotonically increasing with respect to x
diffs = np.diff(y)
is_monotonic_increasing = all(d >= 0 for d in diffs)
print(f"Is y monotonically increasing with x? {is_monotonic_increasing}")


# Check if y is monotonically decreasing with respect to x
is_monotonic_decreasing = all(d <= 0 for d in diffs)
print(f"Is y monotonically decreasing with x? {is_monotonic_decreasing}")
```

```python
# prompt: State which of the following statements are True or False .
#   I. A joint probability is the likelihood of two events occurring together at the same time
#  II. Marginal probability is the probability of the occurrence of two events at the same time

I. True. A joint probability represents the probability of two or more events happening simultaneously.

II. False. Marginal probability is the probability of a single event occurring, irrespective of other events.  It's obtained by summing the joint probabilities over all possible outcomes of the other event(s).
```
