```python
data = pd.read_csv('mtcars2.csv')
data.head()
```

```python
filter_1 = data['cyl'] == 4
print(filter_1.dtype)
```

```python
mask = np.logical_and(data['cyl']> 4, data['hp'] > 100)
data[mask]
```

```python
data.query('(cyl > 4 and hp > 100) | mpg > 20')
```

```python
data.query('mpg > 20')['hp'].mean()
```

```python
data.groupby('cyl')['hp'].mean()
```

```python
data.groupby(['cyl', 'hp'])['mpg'].mean()
```

```python
data.groupby(['cyl', 'hp'])['mpg'].mean().sort_index(ascending=False)
```

```python
data.groupby('cyl').agg(mean_hp=('hp', 'mean'), max_mpg=('mpg', 'max'))
```
