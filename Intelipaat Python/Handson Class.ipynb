# **  Creating Series from Different DS**

```python
import numpy as np
import pandas as pd

# Using the array
array_1 = np.array([1,2,3,4,5])
ser_array = pd.Series(array_1)
ser_array
print(f"Series using array:\n",ser_array)

# Using the list
list_1 = [1,2,3,4,5]
ser_list = pd.Series(list_1)
ser_list
print(f"Series using list:\n",ser_list)


# Using the tuple
tuple_1 = (1,2,3,4,5)
ser_tuple = pd.Series(tuple_1)
ser_tuple
print(f"Series using tuple:\n",ser_tuple)


# Using the set - Set can not be converted into the Series as SET is a Unordered Data Stucture
set_1 = {1,2,3,4,5}
#
# ser_set = pd.Series(list(set_1))
ser_set = pd.Series(np.array((set_1)))
ser_set
print(f"Series using tuple:\n",ser_set)


```
