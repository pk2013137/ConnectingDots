# **Q16 : Write a function to perform the following operation:**
Increase the grade of the house by 1 if the sqft of the house is greater than 0 and less than
equal to 400. If the total sqft of the house is greater than 400, increase the grade of the house
by 2.
After creating the above function, which syntax will be using when using apply function to run
the above created function in the dataset?

```python
def grade_increase(x):
    if x['sqft_living'] > 0 and x['sqft_living'] <= 400:
        x['grade'] += 1
    elif x['sqft_living'] > 400:
        x['grade'] += 2

    return x
```

```python
print(house_ds.head(10))
house_ds_1 = house_ds.apply(grade_increase, axis=1)
print(house_ds_1.head(10))
# house_ds_1 = grade_increase(house_ds)
# Ans data.apply(your_function_name(arg1))
```

**IF WE WANT THE FUNCTION TO WORK ON EVERY ROW THEN WE MUST USE APPLY, OTHERWISE WE CAN DIRECTLY CALL THE FUNCTION BY NAME**

```python
# ============ Approach 2 ==============
def price_inscrease(x):
    if x['sqft_living'] > 0 and x['sqft_living'] <= 400:
        x['grade'] += 1
        return x['grade']
    elif x['sqft_living'] > 400:
        x['grade'] += 2
        return x['grade']

house_ds['grade_new'] = house_ds.apply(price_inscrease, axis=1)

# price_inscrease(house_ds)
house_ds[['grade', 'grade_new', 'sqft_living']].head(20)
```
