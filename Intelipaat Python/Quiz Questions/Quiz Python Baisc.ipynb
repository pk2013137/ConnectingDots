**Question 7** : What is the output of the following code? a = 10 b = 20 print(a > 5 & b > 15)


**& - Bitwise operator** and its precidence with other operator sign

```python
# 10 = (01010)
a = 10
# 20 = (10100)
# 5 =  (00101)
b = 20
print(a & b)
print(a > 5)
print(b > 15)
print(5 & b)
c = (a > 5) & (b > 3)
print("value of c is {}".format(c))
print(a > 5 & b > 15)
print(True & True)
```
