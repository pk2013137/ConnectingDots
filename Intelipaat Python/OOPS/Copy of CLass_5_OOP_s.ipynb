```python
class Student:
    def __init__(self, name, cls, section, marks):
        self.name = name
        self.cls = cls
        self.section = section
        self.marks = marks

    def display(self):
        print(self.name, self.cls, self.section, self.marks)

    def highest(self):
        print(max((a.marks, a.name),
                  (b.marks, b.name),
                  (c.marks, c.name),
                  (d.marks, d.name),
                  (e.marks, e.name)))

a=Student('Ajay', '10th', 'A', 85)
b=Student('Balu', '9th', 'C', 78)
c=Student('Don', '5th', 'B', 93)
d=Student('John', '7th', 'B', 81)
e=Student('David', '6th', 'A', 89)

a.display()
b.display()
c.display()
d.display()
e.display()

a.highest()
```

# Single level inheritance

```python
# class A ----> class B
# parent class A ----> child class B
```

```python
class Parent:
    def __init__(self):
        self.name = 'john'

class child(Parent):
    def display(self):
        print(self.name)


obj = child()
obj.display()
```

```python
class student:
    def __init__(self):
        self.name = 'rohit'
        self.cls = 5
        self.marks = 88

class abc(student):
    def display(self):
        print(self.name, self.cls, self.marks)

obj = abc()
obj.display()
```
