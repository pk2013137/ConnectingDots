```python
class Vehicle:
  def __init__(self):
    self.basefare = 10
    print(f"Initialization from Vehicle class and setting basefare as {self.basefare} for object {self}")

  def fare(self):
    print("it should be implemented in child class")


# Class of vehicle type and specialized fare() method
class Bus(Vehicle):
  def fare(self, busfare):
    self.busfare = busfare
    total_fare = self.basefare + (0.5 * self.busfare)
    print("Bus fare",total_fare)
    return total_fare

class Train(Vehicle):
  def fare(self, trainfare):
    self.trainfare = trainfare
    total_fare = self.basefare + (0.1 * self.trainfare)
    print("Train fare",total_fare)
    return total_fare

class Car(Vehicle):

  def fare(self, carfare):
    self.carfare = carfare
    total_fare = self.basefare + (0.8 * self.carfare)
    print("Car fare",total_fare)
    return total_fare

class Truck(Vehicle):
  def fare(self, carfare):
    self.carfare = carfare
    total_fare = self.basefare + (0.9 * self.carfare)
    print("Truck fare",total_fare)
    return total_fare

class Ship(Vehicle):
  def fare(self, shipfare):
    self.shipfare = shipfare
    total_fare = self.basefare + (1 * self.shipfare)
    print("Ship fare",total_fare)
    return total_fare

# parametized function to take dynamic object and call respective method
def total_fare(vehicle, vehiclefare):
  return vehicle.fare(vehiclefare)

# Creating object for different vehicle type
bus = Bus()
train = Train()
car = Car()
truck = Truck()
Ship = Ship()

# Calculate sum of fare for different vehicle
final_fare = total_fare(bus,20) + total_fare(train,10) + total_fare(car,50)+total_fare(truck,100) + total_fare(Ship,200)
print("Total fare for all vehicles",final_fare)

# 15+11+18+19
```

# Question 3 : Max score function

**Using just Dictionay data structure**

```python
import pandas as pd
import numpy as np
import datetime
from datetime import datetime

def max_score(match_dict):
    # Initialize a dictionary to store total scores of each player
    unique_high_score = {}

    # Loop through each match in the dictionary
    for match, scores in match_dict.items():
        for player, score in scores.items():
            # check if player high score exist in dictonary
          if (player in unique_high_score) and (unique_high_score[player] > score):
                continue
          else:
                unique_high_score[player] = score

    # Find the player with the highest score
    top_player = max(unique_high_score, key=unique_high_score.get)
    top_score = unique_high_score[top_player]
    return (top_player, top_score)

match_dict = {'test1':{'Dhoni':56, 'Balaji' : 85}, 'test2':{'Dhoni': 87, 'Balaji':200}, 'test3':{'Balaji': 80, 'Dravid':105}}
print(max_score(match_dict))
```

# Question 6 : Implementation of abstraction

```python
from abc import ABC, abstractmethod

# Step 1: Create an abstract class
class Shape(ABC):
    @abstractmethod
    def area(self):
        """Calculate area of the shape"""
        pass

    @abstractmethod
    def perimeter(self):
        """Calculate perimeter of the shape"""
        pass

# Step 2: Implement concrete classes inheriting the abstract class
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

    def perimeter(self):
        return 2 * 3.14159 * self.radius

# Step 3: Use the concrete classes
rectangle = Rectangle(5, 10)

print("Rectangle Area:", rectangle.area())
print("Rectangle Perimeter:", rectangle.perimeter())
```

# **Question 7 : Create a program in python to demonstrate Polymorphism.**

```python
class Animal:
    def __init__(self, name):
        self._name = name  # Protected member
        self.__sound = "Generic sound"  # Private member using name mangling

    def make_sound(self):
        """Base method for polymorphism"""
        return f"{self._name} makes {self.__sound}"

class Dog(Animal):
    def __init__(self, name):
        super().__init__(name)
        self.__sound = "Bark"  # Overriding the private member

    def make_sound(self):
        """Override make_sound to demonstrate polymorphism"""
        return f"{self._name} makes {self.__sound}"

class Cat(Animal):
    def __init__(self, name):
        super().__init__(name)
        self.__sound = "Meow"  # Overriding the private member

    def make_sound(self):
        """Override make_sound to demonstrate polymorphism"""
        return f"{self._name} makes {self.__sound}"

# Demonstrate polymorphism
def describe_sound(animal):
    print(animal.make_sound())

# Create objects of Dog and Cat
dog = Dog("Buddy")
cat = Cat("Whiskers")

# Call describe_sound with different objects
describe_sound(dog)  # Polymorphism in action
describe_sound(cat)

# Access protected and private members
print("\nProtected member access:")
print(f"Dog's name: {dog._name}")  # Accessing the protected member

print("\nPrivate member access:")
print(f"Dog's sound using name mangling: {dog._Dog__sound}")  # Accessing private member
print(f"Cat's sound using name mangling: {cat._Cat__sound}")  # Accessing private member
```
