# Multilevel Inheritance

```python
# In multilevel inheritance, child class inherits from the parent class and then  the child class becomes
# a parent class for another child class

class A ----> class B ----> class C
```

```python
class Parent:
    def __init__(self):
        self.name = 'Rohan'
        self.age = 30

class Child(Parent):
    def __init__(self):
       self.age = 40
       Parent.__init__(self)

class Child2(Child):
    def __init__(self):
        self.prof = 'banker'
        Child.__init__(self)


    def display(self):
        print(self.name, self.age, self.prof)


obj = Child2()
obj.display()
```

```python
class Parent:
    def __init__(self):
        self.age = 50

class Child(Parent):
    def __init__(self):
        Parent.__init__(self)
        self.age = 40

class Child2(Child):
    def __init__(self):
        Child.__init__(self)
        self.prof = 'banker'


    def display(self):
        print(self.age, self.prof)


obj = Child2()
obj.display()
```

# Multiple Inheritance

```python
# In multiple inheritance, the child class inherits from more than one parent class
```

```python
class abc:
    def __init__(self):
        self.name = 'Ravindar'

class efg:
    def __init__(self):
        self.age = 40

class child4(abc,efg):
    def __init__(self):
        abc.__init__(self)
        efg.__init__(self)
        self.prof = 'IT'

    def display(self):
        print(self.name, self.age, self.prof)

obj = child4()
obj.display()
```

```python
# WAP to using multiple inheritance,
# take paitent name, paitent address, illness, age, bloodgroup
# and display all of them

class p_patient:
    def __init__(self):
        self.name = 'Lakshmi'

class p_patient1:
    def __init__(self):
        self.address = 'S17,Lakshmi vilas, Bangalore, Karnataka'

class child(p_patient,p_patient1):
    def __init__(self):
        p_patient.__init__(self)
        p_patient1.__init__(self)
        self.illness = 'Fever'
        self.age = 30
        self.bloodgroup = 'A +ve'

    def display(self):
        print (self.name, self.address, self.illness, self.age, self.bloodgroup)

obj = child()
obj.display()
```

```python
class Patient:
    def __init__(Details):
        Details.name = 'Rohan'
        Details.age = 70
class Address:
    def __init__(Details):
        Details.No = 60
        Details.Street = '40 MG Road'
        Details.Postcode = '789056'

class Illness:
    def __init__(Details):
        Details.BloodGroup = 'AB+'


class Condition(Patient, Address, Illness):
    def __init__(Details):
        Patient.__init__(Details)
        Address.__init__(Details)
        Illness.__init__(Details)

    def display(Details):
        print(Details.name, Details.age, Details.No, Details.Street, Details.Postcode, Details.BloodGroup)


obj = Condition()
obj.display()
```

# Hierarchical Inheritance

```python
class Parent:
    def __init__(self):
        self.name = 'John'
        self.age = 25
        self.prof = 'IT'

class Child1(Parent):
    def __init__(self):
        self.city = 'London'
        Parent.__init__(self)

    def display(self):
        print(self.name, self.age,self.city)

class Child2(Parent):
    def __init__(self):
        self.country = 'Uk'
        Parent.__init__(self)

    def display(self):
        print(self.name,self.prof,self.country)

obj1 = Child1()
obj1.display()

obj2 = Child2()
obj2.display()
```

```python
# WAP using Hierarchical inheritance  and you have to displace

# # parent class- company name, company location, , count of employees in the company
# # your name  --> # # display the output
# # your designation --- > # # display the output

class parent:
    def __init__(self):
        self.CompanyName='Google'
        self.CompanyLocation= 'Bangalore'
        self.CountEmp= 150

class child1(parent):
    def __init__(self):
        parent.__init__(self)
        self.name= 'Mak'

    def disp(self):
        print(self.CompanyName, self.CompanyLocation, self.CountEmp,self.name)

class child2(parent):
    def __init__(self):
        parent.__init__(self)
        self.designation= 'DS'

    def display(self):
        print(self.CompanyName, self.CompanyLocation, self.CountEmp,self.designation)

obj1=child1()
obj1.disp()
obj2=child2()
obj2.display()
```

# Polymorphism

```python
- It assures that the data can be displayed in more than one way
```

```python
class parent:
    def __init__(self):
        self.shape = 'Square'
        self.side = 4

    def Area(self):
        print("The area of {} is : {}".format(self.shape, self.side* self.side))

class child:
    def __init__(self):
        self.shape = 'Rectangle'
        self.length = 10
        self.breadth = 5

    def Area(self):
              print("The area of {} is : {}".format(self.shape, self.length * self.breadth))

obj = parent()
obj.Area()

obj1 = child()
obj1.Area()
```

```python
# Abstraction

- It ensures that the internal functioning is hidden from the user
i.e the users knows what the code does, but not how its been done
```

```python
from abc import ABC, abstractmethod

class Sample(ABC):
    @abstractmethod
    def Abstract_method(self):
        print('ABCDE')

class Child(Sample):
    def Abstract_method(self):
        print('Abstraction')

obj = Child()
obj.Abstract_method()
```
