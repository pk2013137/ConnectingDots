# Multi-level Inheritance

```python
class grandparent:

    def __init__(self, fn, mn,av):
        self.father = fn
        self.mother = mn
        self.asset_value =av

    def grand_parent_info(self,fa,ma):
        return f"{self.father} and {self.mother} who are aged at {fa} and {ma} respectively, is leaving behind {self.asset_value} million dollars for this kids"

class parent(grandparent): #python way of inheritance
    pass#to say it is empty doing nothing

class child(parent):
    def __init__(self, cn, mn, fn, av):
        self.child_name = cn
        parent.__init__(self, fn, mn, av)

    def child_info(self, ca):
        return f"{self.child_name} will be inheriting {self.asset_value} millions from his/her partents {self.father} and {self.mother}"
```

```python
ch_ins = child("Alpha", "Gamma", "Beta", 1.4)
print(ch_ins.grand_parent_info(75, 71))
print(ch_ins.child_info(14))
```

# Multiple Inheritance

```python
class grandparent:

    def __init__(self, fn, mn,av):
        self.father = fn
        self.mother = mn
        self.asset_value =av

    def grand_parent_info(self,gfa,gma):
        return f"{self.father} and {self.mother} who are aged at {gfa} and {gma} respectively, is leaving behind {self.asset_value} million dollars for this kids"

class parent: #python way of inheritance

    def parent_info(self, fa, ma):
        return "This is coming from the Parent Class"

class child(grandparent, parent):

    def __init__(self, cn, mn, fn, av):
        self.child_name = cn
        grandparent.__init__(self, fn, mn, av)

    def child_info(self, ca):
        return f"{self.child_name} will be inheriting {self.asset_value} millions from his/her partents {self.father} and {self.mother}"
```

```python
ch_ins = child("Alpha", "Gamma", "Beta", 1.4)
print(ch_ins.grand_parent_info(75, 71))
print(ch_ins.parent_info(45, 41))
print(ch_ins.child_info(14))
```

# Method Overriding

```python
class first:

    def welcome(self, name):
        return f"Welcome to Python Programming, {name}!!!"

class second(first):

    def welcome(self, name, age, sal):
        return f"{name} is making {sal} thousands per month at the age of {age}"
```

```python
ins = second()
print(ins.welcome("Gowtham", 21, 65))
```

```python
fins = first()
fins.welcome("Gowtham")
```

# Encapsulation

#### Public Access Modifier
#### Private Access Modifier
#### Protected Access Modifier

```python
class encap:

    def public_method(self, name):
        return f"Welcome to Public method, {name}"

    def _private_method(self, name):
        return f"Welcome to Private Method, {name}"

    def __protected_method(self, name):
        return f"Welcome to Protected Method, {name}"

class encap_inherited(encap):
    pass
```

```python
ins = encap_inherited()
ins.public_method("Alpha")
ins._private_method("Gamma") # Never Access Variables and Methods starting with an Underscore
```

```python
ins.__protected_method("Gowtham")
```
