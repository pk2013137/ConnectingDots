Join Sets

```python
#union
set1 = {1,2,3,'a','b'}
set2 = {1,2,6,7,'a','c'}

set3 = set1.union(set2)
print(set3)
```

```python
#intersection
set1 = {1,2,3,'a','b'}
set2 = {1,2,6,7,'a','c'}

set3 = set1.intersection(set2)
print(set3)
```

```python
#difference
set1 = {1,2,3,'a','b'}
set2 = {1,2,6,7,'a','c'}
set3 = set1.difference(set2)
print(set3)
```
