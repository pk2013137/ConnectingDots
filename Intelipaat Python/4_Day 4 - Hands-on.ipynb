```python
def armstrong(n):

    total = 0

    for i in str(n):
        print(f"{i} is accessed and it is going to be powered by {len(str(n))}")
        total += int(i) ** len(str(n))

    if n == total:
        return f"{n} is an Armstrong Number"
    else:
        return f"{n} is Not an Armstrong Number"
```

```python
armstrong(153)
```
