```python
# prompt: Which customer(CustomerID) has done highest transaction in the dataset

# Calculate total amount for each transaction
df['TotalAmount'] = df['Quantity'] * df['UnitPrice']

# Group by CustomerID and sum the total amount for each customer
customer_total = df.groupby('CustomerID')['TotalAmount'].sum()

# Find the customer with the highest total transaction amount
customer_with_highest_transaction = customer_total.idxmax()

print(f"The customer with the highest total transaction amount is: {customer_with_highest_transaction}")

customer_total.sort_values(ascending=False)
```

```python
# prompt: Create a DataFrame with unique description of item ,count of the item ,total amount paid for the item from dataframe df. Then get the least purchased item.

# Create a DataFrame with unique descriptions, counts, and total amounts
item_summary = df.groupby('Description').agg({'Quantity': 'sum', 'UnitPrice': 'sum'})
item_summary = item_summary.rename(columns={'Quantity': 'Count', 'UnitPrice': 'TotalAmount'})

# Get the least purchased item
least_purchased_item = item_summary['Count'].idxmin()

print(f"The least purchased item is: {least_purchased_item}")
```
