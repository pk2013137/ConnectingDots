**4. Find out the outliers from the price columns and remove it.**

```python
import seaborn as sns
sns.boxplot(x=df['price'])
```

```python
import numpy as np
q1 = np.percentile(df['price'],25)
q3 = np.percentile(df['price'],75)
print(q1,q3)


iqr = q3 -q1
print(iqr)

ul = q3+iqr*1.5
ll = q3-iqr*1.5

print(ll, ul)
```

```python
df3 = df[(df.price>ll) & (df.price<=ul)]
df3
```

```python
sns.boxplot(x=df3['price'])
```
