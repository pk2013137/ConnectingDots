```python
class Parent:
  def __init__(self) :
    self.name = "Bread"

class child_1(Parent) :
  def __init__(self):
    Parent.__init__(self)
    self.quantity = 10

class child_2(child_1) :
    def __init__(self):
      child_1.__init__(self)
      self.weight = 10

class child_3(child_2) :
    def __init__(self):
      child_2.__init__(self)
      self.price = 10
      self.price = self.weight*self.price*self.quantity


class child_4():
  def __init__(self):
    child_3.__init__(self)
    # self.totalprice = 0
    self.totalprice = self.weight*self.price*self.quantity
  def display(self):
      print(f"{self.name} weight, quantity, price per unit and totalprice is : {self.weight} , {self.quantity}, {self.price}, {self.totalprice}")

obj = child_4()
obj.display()
```

```python
class company:
  def __init__(self) :
    self.company_name = "Verizon"
    self.company_location = "Texas"
    self.countOfEmplyee = 130000


class your_name :
  def __init__(self):
    company.__init__(self)
    self.name = "Prianshu Kumar"
    self.company_location = "Irving"

  def display(self) :
      print(f"Ename : {self.name} , company_name: {self.company_name} , countofEmplyee:{self.countOfEmplyee}, location:{self.company_location}")

class your_degignation :
  def __init__(self):
    company.__init__(self)
    self.designation = "ML Architect"

  def display(self) :
      print(f"company_name: {self.company_name} , countofEmplyee:{self.countOfEmplyee}, location:{self.company_location}, designation: {self.designation}")

obj_1 = your_name()
obj_1.display()

obj_2 = your_degignation()
obj_2.display()
```

```python
#without fuction
user_inp = input("Enter test for palidorme test")
print(type(user_inp))
str_list = list(user_inp)
print(type(str_list))
print(str_list.reverse())
print(str_list[-1])
print(len(str_list))

count_match = 0
for i in range(len(str_list)):
  if str_list[i] == str_list[-1-i] :
      count_match = i+1
      if (count_match == len(str_list)) :
          print("Palindrome")
```
