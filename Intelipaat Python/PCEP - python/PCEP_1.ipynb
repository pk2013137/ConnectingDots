```python
counter = 1
while counter < 11:
  print(counter)
  if counter % 5 == 0:
        print("continue")
        counter += 1
        continue
  elif counter == 9:
        print("pass")
        counter += 1
        pass
  else:
        print("else")
        counter += 1


print("done")
```

```python
user_guessed_year = int(input("When was Python 1.0 released? "))
actual_year = 1994
print(type(actual_year))

while True:

    if user_guessed_year > actual_year :
        print("It was earlier than that!")
        user_guessed_year = int(input("When was Python 1.0 released? "))
        continue
    elif user_guessed_year < actual_year :
        print("It was later than that!")
        user_guessed_year = int(input("When was Python 1.0 released? "))
        continue
    else:
      print("Correct!")
      break
```
