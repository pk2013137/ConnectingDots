```python
spendings = [1346.0, 987.50, 1734.40, 2567.0, 3271.45, 2500.0, 2130.0, 2510.30, 2987.34, 3120.50, 4069.78, 1000.0]
low_spending_count = 0
normal_spending_count = 0
high_spending_count = 0
for spending in spendings:
    if spending < 1000 :
        # print("the spendings were low (< 1000.0)")
        low_spending_count +=1
    elif (spending >= 1000) and (spending <= 2500) :
        # print("the spendings were normal (between 1000.0 and 2500.0 inclusive)")
        normal_spending_count+=1
    else :
        print("the spendings were high (> 2500.0)")
        high_spending_count+=1

print(f"Numbers of months with low spendings: {low_spending_count}, normal spendings: {normal_spending_count}, high spendings: {high_spending_count}.")
```

```python
sample_dict = {
    "mouth": "Mund",
    "finger": "Finger",
    "leg": "Bein",
    "hand": "Hand",
    "face": "Gesicht",
    "nose": "Nase"
}
inp_data = input("Enter a word in English or EXIT: ")

while inp_data != "EXIT" :
    if inp_data in sample_dict :
        print(f"Translation: {sample_dict[inp_data]}")
    else:
      print("No Match!")
    inp_data = input("Enter a word in English or EXIT: ")

else :
    print("Bye!")
```
