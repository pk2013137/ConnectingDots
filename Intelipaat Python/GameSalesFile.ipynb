**Q4. What year experienced the most lucrative sales performance on a global scale?**

```python
df_gbs = df.groupby(by=['YOR'])['GBS'].sum().sort_values(ascending=False).reindex()
y = df_gbs.to_frame()

```

```python
plt.figure(figsize = (15,10))
sns.barplot(x = 'YOR', y = 'GBS', data = df)
plt.xticks(rotation = 90)
plt.show()
```

```python
plt.figure(figsize = (15,10))
sns.barplot(x = 'YOR', y = 'GBS', data = y)
plt.xticks(rotation = 90)
plt.show()
```

**Q5. Which kind of video game consistently garners the highest revenue per unit sold globally?**

```python
df_gbs = df.groupby(by=['Type'])['GBS'].sum().sort_values(ascending=False)
video = df_gbs.to_frame().reset_index()
video
```

```python
plt.figure(figsize = (15,10))
sns.barplot(x = 'Type', y = 'GBS', data = video)
plt.xticks(rotation = 90)
plt.show()
```
