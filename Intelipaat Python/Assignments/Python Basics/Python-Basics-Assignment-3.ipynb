pythagorean_triplet which takes only 1 param i.e N, and determine all 3 possible values of a, b, c
where
1. a**2 + b**2 = c**2
2. a <= N & b <= N & c <= N


```python
def pythagorean_triplet(N):
  for i in range(1, N+1):
    for j in range(1, N+1):
      for k in range(1, N+1):
        # print(i, j, k)
        if (k>j and k>i) and not (i==j==k):
          if(i**2 +j**2 == k**2):
           print(i, j, k)
          else:
           continue

pythagorean_triplet(20)
```
