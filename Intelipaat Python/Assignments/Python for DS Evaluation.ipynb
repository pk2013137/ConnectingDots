```python
# prompt: Using dataframe sp_companies: plot graph number of companies per sector

import altair as alt

# Count the number of companies in each sector
sector_counts = sp_companies['Sector'].value_counts().reset_index()
sector_counts.columns = ['Sector', 'Count']

# Create the bar chart
chart = alt.Chart(sector_counts).mark_bar().encode(
    x='Sector',
    y='Count'
).properties(
    title='Number of Companies per Sector'
)
chart
```

```python
# prompt: plot pie chart for sector and count of company

import matplotlib.pyplot as plt

# Assuming 'sector_counts' DataFrame from the previous code is still available

plt.figure(figsize=(12, 8))
plt.pie(sector_counts['Count'], labels=sector_counts['Sector'], autopct='%1.1f%%', startangle=90)
plt.title('Distribution of Companies across Sectors')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

```python
# prompt: area plot for marketcap vs ebitdata

import matplotlib.pyplot as plt

# Assuming 'sp_companies' DataFrame is already loaded

# Sample data (replace with your actual data)
# Assuming 'MarketCap' and 'EBITDA' are column names in your DataFrame.
# If not, replace them with your actual column names.
market_cap_by_sector = sp_companies.groupby('Sector')['Marketcap'].sum()
ebitda_by_sector = sp_companies.groupby('Sector')['Ebitda'].sum()


# Create the area plot
plt.figure(figsize=(12, 6))
plt.stackplot(market_cap_by_sector.index, market_cap_by_sector.values, ebitda_by_sector.values, labels=['Market Cap', 'EBITDA'])
plt.xlabel('Sector')
plt.ylabel('Value')
plt.title('Market Cap vs EBITDA by Sector')
plt.legend(loc='upper left')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()  # Adjust layout to prevent labels from overlapping
plt.show()
```
