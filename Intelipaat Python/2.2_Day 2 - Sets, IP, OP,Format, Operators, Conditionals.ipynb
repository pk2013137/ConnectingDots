# Assign grades to school students
1. Avg. Marks and Attendance % of the student must be greater than 90 - A
1. Avg. Marks and Attendance % of the student must be greater than 80 - B
1. Avg. Marks and Attendance % of the student must be greater than 70 - C
1. Avg. Marks and Attendance % of the student must be greater than 60 - D
1. Avg. Marks and Attendance % of the student must be greater than 50 - E
1. Avg. Marks and Attendance % of the student must be lesser than 50 - F

```python
student_name = input("Please enter your name: ")
marks_in_subs = list(map(lambda x : int(x), input("Please enter your Marks by 1.Maths | 2.Physics | 3.Chemistry | 4.Language: ").split()))
attendance_pct = round(eval(input("How many classes have you attended out of 180 classes: ")) / 180, 2)*100
```

```python
student_name
```

```python
marks_in_subs
```

```python
avg = round(sum(marks_in_subs) / 400, 2) * 100
avg
```

```python
attendance_pct
```

```python
if attendance_pct >= 90 and avg >= 90:
    print(f"{student_name} is an A grade student with a total aggregate of {avg}% and {attendance_pct}% in attendance")
elif attendance_pct >= 80 and avg >= 80:
    print(f"{student_name} is a B grade student with a total aggregate of {avg}% and {attendance_pct}% in attendance")
elif attendance_pct >= 70 and avg >= 70:
    print(f"{student_name} is a C grade student with a total aggregate of {avg}% and {attendance_pct}% in attendance")
elif attendance_pct >= 60 and avg >= 60:
    print(f"{student_name} is a D grade student with a total aggregate of {avg}% and {attendance_pct}% in attendance")
elif attendance_pct >= 50 and avg >= 50:
    print(f"{student_name} is an E grade student with a total aggregate of {avg}% and {attendance_pct}% in attendance")
else:
    print(f"{student_name} is a F grade student with a total aggregate of {avg}% and {attendance_pct}% in attendance")
```
